```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

t2012 = pd.read_csv("Tabela5-sem_emprego_2012.csv")
t2026 = pd.read_csv("Tabela5-sem_emprego_2026.csv")

t2012 = t2012.rename(columns={"mulheres": "mulheres_2012", "homens": "homens_2012"})
t2026 = t2026.rename(columns={"mulheres": "mulheres_2026", "homens": "homens_2026"})

comp = t2012.merge(t2026, on=["Sigla", "Código", "Estado"], how="inner")

t1_1_1 = pd.read_excel("Tabela 1.1.1.xls", skiprows=4)

t1_1_1 = t1_1_1.dropna(subset=[t1_1_1.columns[0]])
t1_1_1["Código"] = t1_1_1.iloc[:, 0].astype(str).str.extract(r"^(\d{2})")
t1_1_1 = t1_1_1.dropna(subset=["Código"])
t1_1_1["Código"] = t1_1_1["Código"].astype(int)

df_final = comp.merge(t1_1_1, on="Código", how="inner")

ordem = comp.sort_values("mulheres_2026")["Estado"]

longo = comp.melt(
    id_vars=["Sigla", "Código", "Estado"],
    value_vars=["mulheres_2012", "mulheres_2026"],
    var_name="Ano",
    value_name="Participacao_mulheres"
)

longo["Ano"] = longo["Ano"].map({"mulheres_2012": "2012 T1", "mulheres_2026": "2026 T1"})

fig, ax = plt.subplots(figsize=(10, 10))

sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax
)

ax.axvline(50, color="red", linestyle="--", linewidth=1)

ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1")
ax.legend(title="Período")

fig.tight_layout()
plt.show()
```
